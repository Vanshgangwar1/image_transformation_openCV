```python
import cv2
```

```python
image=cv2.imread("vansh_palak.jpg")
# this will load the image 
if image is None:
    print("no image found !!!!!")
# else :
    # print(image)
# this will print image in the form of pixels 

```

```python
# displaying the image 
# 1.imshow - loads the image 
"""with this we have to use waitkey(0) to stop the window 
and then we have to destroy all the windows """

if image is not None:
        
    cv2.imshow("vansh loves palak",image)
    cv2.waitKey(0) 
    # this will pause the window till the user presses any key 
    cv2.destroyAllWindows()
else :
    print("error !!!!!iamge not found ")
```

```python
# saving an image  after making any change 
# imwrite()
if image is not None:
    success=cv2.imwrite("output_image.png",image)
    if success:
        print("image saved successfully")
    else:
        print("failed to save an image")
else:
    print("image not found")
```

```python
#4. image dimensions -

# .shape
"""displays  the height width color channel(blue,green,red) of an image"""
h,w,c=image.shape
# channel(0,0,0) - grayscale
print("height:",h)
print("width:",w)
print("color channels:",c)
```

```python
# grey scale conversion 
# cvtColor 
"""BGR2GRAY , GRAY2BGR , .........."""
if image is not None:
    
    gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
    if gray is not None :
        cv2.imshow("Grayscale image:",gray)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("no gray image found ")
else :
    print("main image not found ")
```

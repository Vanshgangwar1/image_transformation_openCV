```python
import cv2
```

```python
image=cv2.imread("vansh_palak.jpg")
```

```python
if image is not None:
    grey=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
    if grey is not None:
        cv2.imshow("grey image :",grey)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("no grey image found")
else:
    print("no image found")
```

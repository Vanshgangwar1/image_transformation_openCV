```python
# drawing a line on an image 
import cv2
import matplotlib.pyplot as plt

# Load and scale image using OpenCV
img = cv2.imread('vansh_palak.jpg')
scaled_img = cv2.resize(img, None, fx=1.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

# Convert BGR (OpenCV) to RGB (Matplotlib)
img_rgb = cv2.cvtColor(scaled_img, cv2.COLOR_BGR2RGB)

# Display image with X and Y axes
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Scaled Image with Axes")
plt.xlabel("X-axis (Width in pixels)")
plt.ylabel("Y-axis (Height in pixels)")
plt.grid(True, alpha=0.3)  # Optional: adds grid lines over the image
plt.show()
```

```python
if img is not None:
    print("image loaded successfully")
    pt1=(500,600)
    pt2=(2500,600)
    color=(255,0,0)
    thickness=4
    cropped=img[500:2500,200:1600]
    cv2.line(cropped,pt1,pt2,color,thickness)
    cv2.imshow("line drawing",cropped)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# drawing a rectangle in an image using rectangle function 
"""cv2.rectangle(img,pt1,pt2,color,thickness)
pt1=top left corner 
pt2=bottom right corner"""
image=cv2.imread("vansh_palak.jpg")
if image is not None:
    pt1=(500,200)
    pt2=(1000,600)
    color=(0,0,0)
    thickness=4
    cv2.rectangle(image,pt1,pt2,color,thickness)
    cv2.imshow("rectangle image :",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("no image found ")
```

```python
# drawing a circle inside an image 
"""
this user center and radiour 
cv2.circle(image,center,radiour,color,thickness)"""

image=cv2.imread("vansh_palak.jpg")
if image is not None:
    center=(150,150)
    radious=100
    color=(0,0,0)
    thickness=-1
    # when we use thickness =-1 , it fills the circle with the color chosen
    
    cv2.circle(image,center,radious,color,thickness)
    cv2.imshow("rectangle image :",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("no image found ")
```

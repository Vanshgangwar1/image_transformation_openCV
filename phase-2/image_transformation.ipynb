```python
"""resizing and scaling iamge 
we use cv2.resize()"""
import cv2
image=cv2.imread("vansh_palak.jpg")
if image is None:
    print("loading error ")
else:
    print("loaded success")
```

```python
resized=cv2.resize(image,(300,300))
# while resizing firse we use width and then height 
print("resized image")
cv2.imshow("resized image :",resized)
cv2.imwrite("resized.jpg",resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# cropping an image using slicing in openCV
"""ek image ka rectangular part cut karna 
purpose - eg we want to detect just a single face from an image 
,we use this if we don  t want unnecessary portion 

image[start y:end y,start x: end x]
y axis = rows top to bottom
x-axis=columns left to right 


"""

if image is not None:
    cropped=image[100:1000,50:750]
    # this will crop the image according to axis points 
    cv2.imshow("original image :",image)
    cv2.imshow("cropped image:",cropped)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# image rotation and flipping :
"""turning an imagea by some angle fromm center is called iamge rotation

flipping is left to right or top to bottom flip


m=cv2.getRotationMatrix2D(center,angle,scale)
image is rotated by center (tuple of floating numbers (x,y))
to rotate image around its center , we generally do 
                 center =((width/2),height/2)
                 
                 
at some angle (
    positive values specifies counter clockwise rotatiion and 
    negative values specifies clockwise rotn
    
)


and the scale is given 
(
    applied along rotation to make the change in image size 
    (scale=1.0 keeps the original size 
    scale=>1.0 zooms in (enlarges the image ))
    scale<1.0 zooms out (shrinks the image ))
"""

# ROTATION -
if image is not None:
    (h,w)=image.shape[:2]
    center=(w//2,h//2)
    M=cv2.getRotationMatrix2D(center,90,1.0)
    rotated=cv2.warpAffine(image,M,(w,h))
    cv2.imshow("rotated image ------",rotated)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# image flipping ---------------
"""flipped=cv2.flip(image,flipcate)
flipcate=- rotate vertically top -> bottom 
flipcate=1 rotate horizontally left -> right
flipcate=-1 rotate both horizontally and vertically 
"""


if image is not None:
    flipped_hori=cv2.flip(image,1)
    flipped_ver=cv2.flip(image,0)
    flipped_both=cv2.flip(image,-1)
    cv2.imshow("original image ",image[100:1500,200:750])
    cv2.imshow("flipped hori",flipped_hori[100:1500,200:750])
    cv2.imshow("flipped ver",flipped_ver)

    cv2.imshow("flipped both",flipped_both)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
